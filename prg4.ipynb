```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
df=pd.read_csv("melb_data.csv")
print("original Database:")
print(df.head())
```

```python
print("\nShape of dataset:")
print(df.shape)
print("\nColumn names:")
print(df.columns)
print("\nDataset information:")
print("\nStatistical summary:")
print(df.describe())
```

```python
df_mean = df.copy()
mean_building_area = df_mean['BuildingArea'].mean()
df_mean["BuildingArea"] = df_mean["BuildingArea"].fillna(
    mean_building_area
)
print("\nMean Building Area:", mean_building_area)
print("Missing Building values after Mean Imputation:")
print(df_mean["BuildingArea"].isnull().sum())
```

```python
df_median = df.copy()
median_year_built = df_median["YearBuilt"].median()
df_median["YearBuilt"] = df_median["YearBuilt"].fillna(median_year_built)
print("\nMedian Year BUilt:", median_year_built)
print("Missin YearBuilt values after Median Imputation:")
print(df_median["YearBuilt"].isnull().sum())
```

```python
df_ffill=df.copy()
df_ffill["Car"]=df_ffill["Car"].ffill()
comparison = pd.DataFrame({
    "Original BuildingArea": df["BuildingArea"].head(15),
    "Mean Imputed BuildingArea": df_mean["BuildingArea"].head(15),
    "Median Imputed YearBuilt": df_median["YearBuilt"].head(15),
    "Original Car": df["Car"].head(15),
    "Forward Fill Imputed Car": df_ffill["Car"].head(15)
})
print("\nComparison of Imputation Methods:")
print(comparison)
```

```python
df_processed = df.copy()
council_mode = df_processed["CouncilArea"].mode()[0]
df_processed["CouncilArea"] = df_processed["CouncilArea"].fillna(council_mode)
print("\nCouncilArea Mode:", council_mode)
print("Missing Categorical values After Imputation:")
print(df_processed["CouncilArea"].isnull().sum())
```

```python
df_lable = df.copy()
label_encoder = LabelEncoder()
df_lable["CouncilArea_Encoded"] = label_encoder.fit_transform(df_lable["CouncilArea"])
print("\nLabel Encoding of CouncilArea:")
print(df_lable[["CouncilArea", "CouncilArea_Encoded"]].head(10))
```
