Data Visualization using Matplotlib and Seaborn: histograms, box plots, scatter plots, pair plots, and correlation heatmaps.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"Iris.csv")
print("First 5 rows of the dataset:")
print(df.head())
print("\nLast 5 rows of the dataset:")
print(df.tail())
```

```python
print("\nShape of the dataset:")
print(df.shape)
print("\nColumn names:")
print(df.columns)
print("\nDataset information:")
df.info()
```

```python
print(df.isnull().sum())
df = df.dropna()
print(df.isnull().sum())
print(df.duplicated().sum())
df = df.drop_duplicates()
print(df.duplicated().sum())
print(df.describe())
```

```python
sns.set_style("whitegrid")
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="SepalLengthCm", bins=10)
plt.title("Distribution of Sepal Length")

# Add X-axis label
plt.xlabel("Sepal Length (cm)")

# Add Y-axis label
plt.ylabel("Number of Flowers")

# Display graph
plt.show()
```

```python
# Create graph area
plt.figure(figsize=(8, 5))

# Create box plot
sns.boxplot(data=df, x="Species", y="SepalLengthCm")

# Add title
plt.title("Sepal Length by Species")

# Add X-axis label
plt.xlabel("Species")

# Add Y-axis label
plt.ylabel("Sepal Length (cm)")
# Display graph
plt.show()
```

```python
# Create graph area
plt.figure(figsize=(8, 5))

# Create scatter plot
sns.scatterplot(
    data=df,
    x="SepalLengthCm",
    y="PetalLengthCm",
    hue="Species"
)
# Add title
plt.title("Sepal Length vs Petal Length")

# Add X-axis label
plt.xlabel("Sepal Length (cm)")

# Add Y-axis label
plt.ylabel("Petal Length (cm)")

# Display graph
plt.show()
```

```python
# Create pair plot
sns.pairplot(df, hue="Species")

# Display graph
plt.show()

```

```python
# Select numerical columns
numeric_columns = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

# Calculate correlation
correlation = df[numeric_columns].corr()

# Display correlation matrix
print(correlation)

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```
